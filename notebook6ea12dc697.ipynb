```python
# A3 — Prétraitement et entraînement

import pandas as pd
from IPython.display import display
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# 1. Charger les 30 caractéristiques et la cible
dataset = load_breast_cancer()

X = pd.DataFrame(
    dataset.data,
    columns=dataset.feature_names
)

# Convention conservée : 0 = malin ; 1 = bénin
y = pd.Series(dataset.target, name="target")

print("Dimensions des données :", X.shape)
print("Valeurs manquantes :", int(X.isna().sum().sum()))
print("Codage : 0 = malin ; 1 = bénin")

# 2. Séparer les données : environ 60 % / 20 % / 20 %
# Le test est mis de côté avant l'entraînement.
X_dev, X_test, y_dev, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

# 25 % des 80 % restants correspondent à environ 20 % du total.
X_train, X_val, y_train, y_val = train_test_split(
    X_dev,
    y_dev,
    test_size=0.25,
    stratify=y_dev,
    random_state=42
)

# 3. Afficher la taille et les classes de chaque ensemble
bilan_decoupage = pd.DataFrame([
    {
        "Ensemble": nom,
        "Observations": len(cible),
        "Cas malins": int((cible == 0).sum()),
        "Cas bénins": int((cible == 1).sum()),
        "Part du total (%)": round(100 * len(cible) / len(y), 2)
    }
    for nom, cible in [
        ("Entraînement", y_train),
        ("Validation", y_val),
        ("Test", y_test)
    ]
])

print("\nDécoupage des données :")
display(bilan_decoupage)

# 4. Construire le pipeline
# Les paramètres du prétraitement seront appris uniquement sur train.
modele = Pipeline([
    ("imputation", SimpleImputer(strategy="median")),
    ("normalisation", StandardScaler()),
    ("classification", LogisticRegression(
        max_iter=3000,
        random_state=42
    ))
])

# 5. Entraîner uniquement sur l'ensemble d'entraînement
modele.fit(X_train, y_train)

print("\nEntraînement terminé.")
print("Modèle : régression logistique")
print("Prétraitement : imputation médiane + standardisation")
print("Nombre de caractéristiques :", modele.n_features_in_)
print("Classes du modèle :", modele.classes_)

# 6. Sauvegarder le bilan du découpage
bilan_decoupage.to_csv(
    "/kaggle/working/A3_decoupage_donnees.csv",
    index=False
)

display(modele)
```

```python
# A4 — Évaluation du modèle

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# 1. Vérifier que la cellule A3 a été exécutée
variables_requises = [
    "modele", "X_val", "y_val", "X_test", "y_test"
]

absentes = [
    nom for nom in variables_requises
    if nom not in globals()
]

if absentes:
    raise RuntimeError(
        "Exécuter d'abord la cellule A3. "
        f"Variables absentes : {absentes}"
    )

# 2. Prédire sans réentraîner le modèle
pred_val = modele.predict(X_val)
pred_test = modele.predict(X_test)

# 3. Calculer les métriques
# IMPORTANT : la classe positive est la classe maligne, codée 0.
def calculer_metriques(nom, y_reel, y_predit):
    return {
        "Ensemble": nom,
        "Exactitude (%)": 100 * accuracy_score(
            y_reel, y_predit
        ),
        "Précision malin (%)": 100 * precision_score(
            y_reel, y_predit,
            pos_label=0,
            zero_division=0
        ),
        "Rappel malin (%)": 100 * recall_score(
            y_reel, y_predit,
            pos_label=0,
            zero_division=0
        ),
        "F1 malin (%)": 100 * f1_score(
            y_reel, y_predit,
            pos_label=0,
            zero_division=0
        )
    }

resultats_evaluation = pd.DataFrame([
    calculer_metriques("Validation", y_val, pred_val),
    calculer_metriques("Test", y_test, pred_test)
])

print("Métriques — classe positive : malin (0)")
display(resultats_evaluation.round(2))

# 4. Afficher les résultats détaillés par classe sur le test
print("\nRapport de classification — ensemble de test :")

print(classification_report(
    y_test,
    pred_test,
    labels=[0, 1],
    target_names=["Malin", "Bénin"],
    digits=4,
    zero_division=0
))

# 5. Construire la matrice de confusion
# Ordre des lignes et colonnes : Malin (0), Bénin (1)
matrice = confusion_matrix(
    y_test,
    pred_test,
    labels=[0, 1]
)

fig, ax = plt.subplots(figsize=(6, 5))

affichage = ConfusionMatrixDisplay(
    confusion_matrix=matrice,
    display_labels=["Malin", "Bénin"]
)

affichage.plot(
    ax=ax,
    cmap="Blues",
    values_format="d",
    colorbar=False
)

ax.set_title("Matrice de confusion — ensemble de test")
ax.set_xlabel("Classe prédite")
ax.set_ylabel("Classe réelle")

plt.tight_layout()

fig.savefig(
    "/kaggle/working/A4_matrice_confusion.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()

# 6. Identifier les erreurs en considérant « malin » comme positif
# Attention : cet ordre diffère d'une matrice où positif = 1.
VP = int(matrice[0, 0])  # Malin prédit malin
FN = int(matrice[0, 1])  # Malin prédit bénin
FP = int(matrice[1, 0])  # Bénin prédit malin
VN = int(matrice[1, 1])  # Bénin prédit bénin

rappel_malin = recall_score(
    y_test,
    pred_test,
    pos_label=0,
    zero_division=0
)

print("\nAnalyse des résultats sur le test :")
print(f"Vrais positifs — malins détectés : {VP}")
print(f"Faux négatifs — malins classés bénins : {FN}")
print(f"Faux positifs — bénins classés malins : {FP}")
print(f"Vrais négatifs — bénins reconnus : {VN}")

print(
    f"\nRappel malin = VP / (VP + FN)"
    f" = {VP} / {VP + FN}"
    f" = {100 * rappel_malin:.2f} %"
)

print(
    f"\nSur {VP + FN} cas malins du test, "
    f"{VP} sont correctement détectés "
    f"et {FN} sont classés bénins."
)

print(
    "Un faux négatif peut entraîner une fausse réassurance. "
    "Une prédiction bénigne ne constitue donc pas "
    "une exclusion certaine de malignité."
)

print(
    "Ces résultats sont une évaluation interne sur un jeu public, "
    "pas une validation pour un usage clinique."
)

# 7. Sauvegarder les résultats
resultats_evaluation.to_csv(
    "/kaggle/working/A4_metriques.csv",
    index=False
)

pd.DataFrame(
    matrice,
    index=["Réel malin", "Réel bénin"],
    columns=["Prédit malin", "Prédit bénin"]
).to_csv(
    "/kaggle/working/A4_matrice_confusion.csv"
)
```

```python
# A5.1 — Sauvegarder et recharger le modèle

from pathlib import Path
import joblib
import numpy as np
import pandas as pd

# Vérifier que A3 a été exécutée
if "modele" not in globals() or "X_train" not in globals():
    raise RuntimeError("Exécuter d'abord la cellule A3.")

dossier_sortie = Path("/kaggle/working")
dossier_sortie.mkdir(parents=True, exist_ok=True)

chemin_modele = dossier_sortie / "modele_cancer_sein.pkl"

# Sauvegarder le pipeline ET le contrat d'entrée
artefact = {
    "pipeline": modele,
    "colonnes": list(X_train.columns),
    "classes": {0: "Malin", 1: "Bénin"}
}

joblib.dump(artefact, chemin_modele)

print("Modèle sauvegardé :", chemin_modele)
print("Taille du fichier :", chemin_modele.stat().st_size, "octets")

# Recharger uniquement un fichier de confiance
modele_charge = joblib.load(chemin_modele)

print("Rechargement réussi.")
print("Nombre de caractéristiques :", len(modele_charge["colonnes"]))
print("Classes :", modele_charge["classes"])
```

```python
# A5.2 — Fonction predire(entree)

def predire(entree):
    """Prédire la classe d'une seule observation."""

    # 1. Convertir l'entrée en tableau
    if isinstance(entree, dict):
        ligne = pd.DataFrame([entree])

    elif isinstance(entree, pd.DataFrame):
        ligne = entree.copy()

    else:
        raise TypeError(
            "L'entrée doit être un dictionnaire "
            "ou un DataFrame d'une seule ligne."
        )

    if len(ligne) != 1:
        raise ValueError("Fournir exactement une observation.")

    if ligne.columns.duplicated().any():
        raise ValueError("Les noms de colonnes doivent être uniques.")

    # 2. Contrôler les colonnes
    colonnes = modele_charge["colonnes"]

    absentes = sorted(set(colonnes) - set(ligne.columns))
    supplementaires = sorted(set(ligne.columns) - set(colonnes))

    if absentes or supplementaires:
        raise ValueError(
            f"Colonnes absentes : {absentes}\n"
            f"Colonnes supplémentaires : {supplementaires}"
        )

    # 3. Respecter l'ordre des variables et vérifier les valeurs
    ligne = ligne.loc[:, colonnes]
    ligne = ligne.apply(pd.to_numeric, errors="raise")

    if not np.isfinite(ligne.to_numpy(dtype=float)).all():
        raise ValueError(
            "Valeurs manquantes ou infinies : corriger les données."
        )

    # 4. Appliquer le pipeline sauvegardé
    pipeline = modele_charge["pipeline"]

    code_classe = int(pipeline.predict(ligne)[0])
    probabilites = pipeline.predict_proba(ligne)[0]

    # Associer les scores aux classes sans supposer leur ordre
    scores = {
        int(classe): float(score)
        for classe, score in zip(
            pipeline.classes_,
            probabilites
        )
    }

    # 5. Retourner la classe et les scores
    return {
        "classe": modele_charge["classes"][code_classe],
        "probabilite": scores[code_classe],
        "probabilite_malin": scores[0],
        "avertissement": (
            "Score du modèle, non une certitude médicale. "
            "La décision finale appartient au professionnel."
        )
    }
```

```python
# A5.3 — Tester une observation après rechargement

import json

if "X_test" not in globals() or "y_test" not in globals():
    raise RuntimeError("Exécuter d'abord la cellule A3.")

# Une observation de démonstration : uniquement les 30 caractéristiques
exemple = X_test.iloc[[0]].copy()

resultat = predire(exemple)

print("Dossier de démonstration : DEMO_001")
print("Classe proposée :", resultat["classe"])

print(
    "Score pour la classe proposée :",
    f"{100 * resultat['probabilite']:.2f} %"
)

print(
    "Score de la classe maligne :",
    f"{100 * resultat['probabilite_malin']:.2f} %"
)

print("Avertissement :", resultat["avertissement"])

# Comparaison pédagogique avec l'étiquette connue du dataset
classe_reelle = modele_charge["classes"][int(y_test.iloc[0])]
print("\nClasse réelle dans le dataset :", classe_reelle)

# Vérifier que la sauvegarde conserve la prédiction
prediction_originale = int(modele.predict(exemple)[0])
prediction_rechargee = int(
    modele_charge["pipeline"].predict(exemple)[0]
)

assert prediction_originale == prediction_rechargee
print("Vérification : même classe avant et après sauvegarde.")

# Sauvegarder l'exemple et la réponse
exemple.to_csv(
    dossier_sortie / "A5_exemple_entree.csv",
    index=False
)

with open(
    dossier_sortie / "A5_exemple_prediction.json",
    "w",
    encoding="utf-8"
) as fichier:
    json.dump(resultat, fichier, ensure_ascii=False, indent=2)
```

```python
# A6 — Temps de réponse et comparaison KLM

import time
import numpy as np
import pandas as pd
from IPython.display import display
from pathlib import Path

# 1. Vérifier que A5 a été exécutée
if "predire" not in globals() or "exemple" not in globals():
    raise RuntimeError(
        "Exécuter d'abord les cellules A5.1, A5.2 et A5.3."
    )

# 2. Échauffement : appels exclus des mesures
for _ in range(20):
    predire(exemple)

# 3. Mesurer la durée de 500 prédictions individuelles
N = 500
durees = []

for _ in range(N):
    debut = time.perf_counter()
    predire(exemple)
    fin = time.perf_counter()

    durees.append(fin - debut)

durees = np.array(durees)

# R est exprimé en secondes
R = float(durees.mean())

print("MESURE DU TEMPS DE PRÉDICTION")
print("Nombre d'appels mesurés :", N)
print(f"Temps moyen R : {R:.6f} s")
print(f"Temps moyen : {1000 * R:.3f} ms")
print(f"Médiane : {np.median(durees):.6f} s")
print(f"Écart-type : {np.std(durees, ddof=1):.6f} s")

print(
    "\nPérimètre : contrôle d'une observation, "
    "prétraitement et prédiction via predire()."
)
print(
    "Hors chargement du modèle depuis le disque, "
    "réseau et affichage dans l'interface."
)

# 4. Valeurs des opérateurs KLM, en secondes
K = 0.28  # Clic ou frappe
P = 1.10  # Pointage
H = 0.40  # Transfert de la main
M = 1.35  # Préparation mentale

# Hypothèse : main déjà sur la souris.
# Import, sélection et vérification du dossier sont communs
# aux deux variantes et exclus de cette comparaison.
#
# A : bouton Analyser -> M + P + K + R
# B : lancement automatique -> R
#
# H = 0 occurrence ; D = 0 occurrence.

temps_A = M + P + K + R
temps_B = R
gain = temps_A - temps_B

# 5. Afficher la comparaison
comparaison_klm = pd.DataFrame({
    "Variante": [
        "A — Bouton Analyser",
        "B — Analyse automatique"
    ],
    "Séquence": [
        "M + P + K + R",
        "R"
    ],
    "Temps estimé (s)": [
        temps_A,
        temps_B
    ]
})

print("\nCOMPARAISON KLM")
display(comparaison_klm.round(6))

print(f"T_A = 1,35 + 1,10 + 0,28 + R = {temps_A:.6f} s")
print(f"T_B = R = {temps_B:.6f} s")
print(f"Gain estimé de B : {gain:.2f} s par déclenchement")

print(
    "\nChoix de conception retenu : variante A, "
    "pour conserver un lancement explicite après "
    "vérification du dossier."
)

print(
    "Limite : KLM décrit une procédure connue, "
    "réalisée par un utilisateur expert sans erreur. "
    "La lecture du résultat et la décision médicale "
    "ne sont pas incluses dans ce calcul."
)

# 6. Sauvegarder les mesures et les calculs
sortie = Path("/kaggle/working")

pd.DataFrame({
    "Appel": np.arange(1, N + 1),
    "Duree_secondes": durees
}).to_csv(
    sortie / "A6_durees_prediction.csv",
    index=False
)

comparaison_klm.to_csv(
    sortie / "A6_comparaison_KLM.csv",
    index=False
)

pd.DataFrame([{
    "Nombre_appels": N,
    "R_moyen_secondes": R,
    "Mediane_secondes": float(np.median(durees)),
    "Ecart_type_secondes": float(np.std(durees, ddof=1)),
    "Temps_A_secondes": temps_A,
    "Temps_B_secondes": temps_B,
    "Gain_secondes": gain
}]).to_csv(
    sortie / "A6_resume.csv",
    index=False
)

print("\nMesures et calculs sauvegardés dans /kaggle/working.")
```
